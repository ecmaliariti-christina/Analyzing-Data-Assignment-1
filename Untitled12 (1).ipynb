```python
#PART 1
#1.1 Sentence Splitting & Tokenization

import nltk
import spacy
import matplotlib.pyplot as plt
from collections import Counter
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import PorterStemmer, LancasterStemmer
import os

# Download required resources 
nltk.download('punkt')

nltk.download('punkt_tab')

# Load spaCy model 
nlp = spacy.load("en_core_web_sm")

# File paths
files = [
    "01.txt",
    "02.txt",
    "03.txt",
    "04.txt",
    "05.txt"
]

stories = {}

# Read files and lowercase
for i, file in enumerate(files):
    with open(file, "r", encoding="utf-8") as f:
        text = f.read().lower()
        stories[f"story_{i+1}"] = text
```

```python
#1.3 NLTK Tokenization

nltk_tokens = {}
nltk_sentences = {}

for name, text in stories.items():
    sentences = sent_tokenize(text)
    words = word_tokenize(text)
    
    nltk_sentences[name] = sentences
    nltk_tokens[name] = words
```

```python
# 1.4 spacy tokenization

spacy_tokens = {}
spacy_sentences = {}

for name, text in stories.items():
    doc = nlp(text)
    
    spacy_sentences[name] = [sent.text for sent in doc.sents]
    spacy_tokens[name] = [token.text for token in doc]
```

```python
# 2.1 Word frequency statistic, per story

freq_per_story = {}

for name, tokens in nltk_tokens.items():
    freq_per_story[name] = dict(Counter(tokens))

print(freq_per_story)
```

```python
# 2.2 total corpus frequency

all_tokens = []

for tokens in nltk_tokens.values():
    all_tokens.extend(tokens)

total_freq = dict(Counter(all_tokens))

print(total_freq)
```

```python
# 3.1 25 most common words, per story

for name, tokens in nltk_tokens.items():
    counter = Counter(tokens)
    most_common = counter.most_common(25)
    
    words = [w[0] for w in most_common]
    counts = [w[1] for w in most_common]
    
    plt.figure()
    plt.bar(words, counts)
    plt.xticks(rotation=90)
    plt.title(f"Top 25 Words - {name}")
    plt.show()
```

```python
# 3.2 25 most common words, total corpus

counter_total = Counter(all_tokens)
most_common_total = counter_total.most_common(25)

words = [w[0] for w in most_common_total]
counts = [w[1] for w in most_common_total]

plt.figure()
plt.bar(words, counts)
plt.xticks(rotation=90)
plt.title("Top 25 Words - Total Corpus")
plt.show()
```

```python
# 4.1 Stemming

porter = PorterStemmer()
lancaster = LancasterStemmer()

porter_tokens = {}
lancaster_tokens = {}

for name, tokens in nltk_tokens.items():
    porter_tokens[name] = [porter.stem(word) for word in tokens]
    lancaster_tokens[name] = [lancaster.stem(word) for word in tokens]
```

```python
# 4.2 Porter

porter_freq = {}

for name, tokens in porter_tokens.items():
    porter_freq[name] = dict(Counter(tokens))
```

```python
# 4.3 Lancaster

lancaster_freq = {}

for name, tokens in lancaster_tokens.items():
    lancaster_freq[name] = dict(Counter(tokens))
```

```python
# 5.1 Cleaning

import re

def clean_gutenberg(text):
    # Remove everything before START marker
    start_pattern = r"\*\*\* START OF THE PROJECT GUTENBERG EBOOK.*?\*\*\*"
    end_pattern = r"\*\*\* END OF THE PROJECT GUTENBERG EBOOK.*?\*\*\*"
    
    text = re.split(start_pattern, text, flags=re.DOTALL)[-1]
    text = re.split(end_pattern, text, flags=re.DOTALL)[0]
    
    return text.strip()
```

```python
def run_ner(sentence):
    doc = nlp(sentence)
    for ent in doc.ents:
        print(ent.text, "->", ent.label_)
```
